```python
df['Order Day'] = df['Order Date'].dt.day_name() #Узнаем название дня
df['Order Month'] = df['Order Date'].dt.month_name() #Узнаем название месяца

season_name = {
    1: 'Winter', 2: 'Winter', 3: 'Spring',
    4: 'Spring', 5: 'Spring', 6: 'Summer',
    7: 'Summer', 8: 'Summer', 9: 'Fall',
    10: 'Fall', 11: 'Fall', 12: 'Winter'
}

df['Order Season'] = df['Order Date'].dt.month.map(season_name)

df['Weekend'] = df['Order Day'].apply(lambda x: 0 if x=='Sunday' or x=='Saturday' else 1)

dummy_cat = ['Ship Mode', 'Segment', 'City', 
             'State', 'Region', 'Category', 
             'Sub-Category', 'Order Day', 
             'Order Month', 'Weekend', 'Order Season', 
             'Product ID']

for col in dummy_cat:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col].astype(str))

df = df.sort_values('Order Date')

df['lag_1']  = df['Sales'].shift(1)
df['lag_7']  = df['Sales'].shift(7)
df['lag_30']  = df['Sales'].shift(30)

df["lag_7_ID"] = df.groupby("Customer ID")["Sales"].shift(7)

df['sales_roll_7'] = df['Sales'].rolling(7).mean()
df['sales_roll_30'] = df['Sales'].rolling(30).mean()

df['roll_std_7'] = df['Sales'].rolling(7).std()

df['growth_1'] = df['Sales'] / df['lag_1']
df['growth_7'] = df['Sales'] / df['lag_7']

df['avg_sales_category'] = df.groupby('Category')['Sales'].transform('mean')
df['avg_sales_city'] = df.groupby('City')['Sales'].transform('mean')
df['avg_sales_customer'] = df.groupby('Customer ID')['Sales'].transform('mean')
df['avg_sales_month'] = df.groupby('Order Month')['Sales'].transform('mean')
df['avg_sales_season'] = df.groupby('Order Season')['Sales'].transform('mean')
df['avg_sales_region'] = df.groupby('Region')['Sales'].transform('mean')
df['avg_sales_segment'] = df.groupby('Segment')['Sales'].transform('mean')
```

```python
df_1 = df.drop(['Row ID', 'Order ID', 'Order Date', 'Ship Date', 'Customer ID', 
         'Customer Name', 'Country', 'Postal Code', 'Product Name', 'Month Year'],
        axis=1)

X = df_1.drop(['Sales'], axis=1)
    
y = df_1['Sales']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=False
)
```

```python
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
model = XGBRegressor(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=6,
    verbosity=1,
    early_stopping_rounds=40,
    eval_metric='rmse'
)

model.fit(
    X_train, y_train,
    eval_set=[(X_test, y_test)]
)
```

```python
y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"MAE  = {mae:.4f}")
print(f"MSE  = {mse:.4f}")
print(f"RMSE = {rmse:.4f}")
print(f"R²   = {r2:.4f}")

```
